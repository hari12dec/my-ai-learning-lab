```python
import pandas as pd
import numpy as np
dataset = pd.read_csv("Placement.csv")
```

```python
from Univariate import Univariate
obj = Univariate()

quan, qual = obj.quanQual(dataset)
```

```python
print(quan)
print(qual)
dataset['ssc_p'].mean()
```

```python
descriptive=pd.DataFrame(index=["Mean", "Median", "Mode"], columns=quan)
descriptive
```

```python
descriptive=pd.DataFrame(index=["Mean", "Median", "Mode", "Q1:25%", "Q2:50%", "Q3:75%", "Q4:100%", "IQR", "1.5rule", "Lesser", "Greater", "Min", "Max", "LowOutLier", "HighOutLier"], columns=quan)
for columnName in quan:
    descriptive.loc["Mean", columnName]=dataset[columnName].mean()
    descriptive.loc["Median", columnName]=dataset[columnName].median()
    descriptive.loc["Mode", columnName]=dataset[columnName].mode()[0]
    descriptive.loc["Q1:25%", columnName]=dataset[columnName].quantile(0.25)
    descriptive.loc["Q2:50%", columnName]=dataset[columnName].quantile(0.50)
    descriptive.loc["Q3:75%", columnName]=dataset[columnName].quantile(0.75)
    descriptive.loc["Q4:100%", columnName]=dataset[columnName].quantile(1)
    descriptive.loc["IQR", columnName]=dataset[columnName].quantile(0.75)-dataset[columnName].quantile(0.25)
    descriptive.loc["1.5rule", columnName]=1.5*descriptive.loc["IQR", columnName]
    descriptive.loc["Lesser", columnName]=dataset[columnName].quantile(0.25)-descriptive.loc["1.5rule", columnName]
    descriptive.loc["Greater", columnName]=dataset[columnName].quantile(0.75)+descriptive.loc["1.5rule", columnName]
    descriptive.loc["Min", columnName]=dataset[columnName].min()
    descriptive.loc["Max", columnName]=dataset[columnName].max()
    descriptive.loc["LowOutLier", columnName]=dataset[columnName].min()<descriptive.loc["Lesser", columnName]
    descriptive.loc["HighOutLier", columnName]=dataset[columnName].max()>descriptive.loc["Greater", columnName]

descriptive
```

```python
for columnName in quan:
    
    lesser=descriptive.loc["Lesser",columnName]
    greater=descriptive.loc["Greater",columnName]
    
    dataset[columnName]=dataset[columnName].clip(lower=lesser, upper=greater)
```
